```python
import pandas as pd
import json
```

```python
results = pd.DataFrame(json.loads(open("../nl_testgeneration_framework/test_outputs/ae7d024d800f4dfdac8153683f958c95/all_results.json").read()))
```

```python
stat_mapping = json.loads(open("../nl_testgeneration_framework/stat_mapping/MBB/Player.json").read())
```

```python
stat_mapping = {x["stat"].lower() : x for x in stat_mapping}
```

```python
import re
```

```python
stat_pattern = re.compile("(s_[a-z0-9_]+)(\s|$)")
```

```python
stat_pattern.findall("s_fieid_goal_made_longest and s_rushing_yards_net")
```

```python
new_recs = [] 
for i,x in results.iterrows():
   
    
    #ground_stat = x["ground_truth_aql"]["conditions"].lower().replace("is not null", "").strip()
    ground_conditions = x["ground_truth_aql"]["conditions"].lower().replace("<=","<").replace(">=",">")
    ground_stats = [t[0] for t in stat_pattern.findall(ground_conditions)]
    
    x["ground_stats"] = ground_stats
    
    x["pred_conditions"] =  x["predicted_aql"]["conditions"].lower()

    
    pred_stats =  stat_pattern.findall(x["predicted_aql"]["conditions"].lower())
    x["pred_stats"] = [t[0] for t in pred_stats]
    
    x["ground_qtype"] =  x["ground_truth_aql"]["query_type"].lower()
    x["pred_qtype"] =  x["predicted_aql"]["query_type"].lower()
    
    if x["ground_stats"] != x["pred_stats"]:
        x["stat_matched"] = False
    else:
        x["stat_matched"] = True
    
    if x["ground_qtype"] != x["pred_qtype"]:
        x["qtype_matched"] = False
    else:
        x["qtype_matched"] = True
        
    x["categories"] = [stat_mapping[y]["category"] for y in ground_stats]
    # x["sub_categories"] = [stat_mapping[y]["sub_category"] for y in ground_stats]
    
    new_recs.append(x)
```

```python
new_recs = pd.DataFrame(new_recs)
```

```python
export = new_recs[["query",  "categories", "ground_stats", "pred_conditions", "pred_stats", "ground_qtype", "pred_qtype", "stat_matched", "qtype_matched"]]
```

```python
true_percentage = len(export[export["stat_matched"] == True]) / len(export)
print(true_percentage)
```
